```python
# TODO - Criando a tabela bronze do volume amazon para o volume amazon_files
# schema = "files.amazon_files"

spark.sql("CREATE SCHEMA IF NOT EXISTS files.amazon_files")

df = spark.read.csv(
    "/Volumes/files/amazon_files/volumes/amazon.csv",
    header=True,
    inferSchema=True
)

df.write.format("delta").mode("overwrite").saveAsTable("files.amazon_files.amazon_bronze")
```

```python
import pyspark.sql.functions as F

spark.sql("SELECT * FROM files.amazon_files.amazon_bronze") \
    .filter(~F.col("rating").rlike(r"^[0-9]\.[0-9]$")) \
        .count()
```
